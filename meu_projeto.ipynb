

# Projeto Prático - Automatização de Movimentação de Arquivos



O desafio consiste em otimizar a rotina de organização de relatórios em Excel, evitando a verificação e a transferência manual de arquivos finalizados para pastas de armazenamento. O objetivo foi criar uma rotina automatizada em Python que identifica arquivos processados, move-os para o diretório correto e registra suas propriedades principais.


* Desenvolver um script em Python para automação da movimentação de arquivos Excel processados.

* Analisar arquivos no diretório de relatórios

* Selecionar arquivos com sufixo "_processado" em seus nomes

* Mover arquivos selecionados para "relatorios processados"

* Utilizar as bibliotecas **os** e **shutil** para interagir com o sistema de arquivos.

* Listar e filtrar arquivos usando estruturas de dados aprendidas anteriormente (listas, estruturas de repetição, estruturas de decisão, arquivos)

* Mover arquivos filtrados usando funções do pacote os e shutil

* Providenciar feedback: informar quantos arquivos foram movidos

```python
# Importação das bibliotecas

import os
import shutil
import time
```

```python
# Definição da pasta de origem onde os relatórios estão salvos
origem = "./relatorios"
```

```python
# Definição da pasta de destino para os arquivos processados
destino = "./relatorios processados"
```

```python
# Garante a existência do pasta de destino
if not os.path.exists(destino):
    os.mkdir(destino)
```

```python
# Lista todos os arquivos presentes na pasta de origem
arquivos = os.listdir(origem)
```

```python
# Dicionário para armazenar as propriedades de cada arquivo
info_arquivos = {}
```

```python
# Filtra apenas os relatórios em Excel que terminam com '_processado'

arquivos_processados = [arquivo for arquivo in arquivos if arquivo.endswith('_processado.xlsx')]

arquivos_processados
```

```python
# Move os arquivos filtrados para a pasta de destino

for arquivo_processado in arquivos_processados:
    caminho_completo_origem = os.path.join(origem, arquivo_processado)
    caminho_completo_destino = os.path.join(destino, arquivo_processado)

    shutil.move(caminho_completo_origem, caminho_completo_destino)

    print(f'Arquivo {caminho_completo_origem} movido para {caminho_completo_destino}')
```

```python
# Coleta o tamanho e a data de criação de cada arquivo e salva no dicionário

for arquivo in arquivos_processados:
    caminho_completo = os.path.join(destino, arquivo)
    
    # Obtendo informações sobre o arquivo
    tamanho = os.path.getsize(caminho_completo)
    data_criacao = time.ctime(os.path.getctime(caminho_completo))
    
    # Adicionando informações no dicionário
    info_arquivos[arquivo] = [data_criacao, tamanho]

info_arquivos
```
